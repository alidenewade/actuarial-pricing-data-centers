# 03 — Copula dependence: Gaussian vs Gumbel

Illustrates the difference in **upper-tail dependence**: Gaussian has $\lambda_U = 0$ for any $\rho < 1$, whereas Gumbel($\theta$) gives $\lambda_U = 2 - 2^{1/\theta} > 0$ for $\theta > 1$.

```python
import numpy as np
import matplotlib.pyplot as plt

from dcrisk.copula.gaussian import GaussianCopula
from dcrisk.copula.gumbel import GumbelCopula

rng = np.random.default_rng(0)

gauss = GaussianCopula(rho=0.7)
gumb  = GumbelCopula(theta=2.0)

U_g = gauss.sample(5000, rng)
U_u = gumb.sample(5000, rng)

print('Gaussian upper-tail dependence (closed-form) =', gauss.upper_tail_dependence())
print('Gumbel   upper-tail dependence (closed-form) =', gumb.upper_tail_dependence())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4.5), sharex=True, sharey=True)
for ax, U, title in [(axes[0], U_g, 'Gaussian, rho=0.7'),
                     (axes[1], U_u, 'Gumbel, theta=2.0')]:
    ax.scatter(U[:, 0], U[:, 1], s=4, alpha=0.4)
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.set_aspect('equal')
    ax.set_title(title)
    ax.set_xlabel('U_1'); ax.set_ylabel('U_2')
    ax.grid(alpha=0.3)
fig.suptitle('Copula scatterplots: note the joint upper-tail clustering of the Gumbel sample')
fig.tight_layout()
```
