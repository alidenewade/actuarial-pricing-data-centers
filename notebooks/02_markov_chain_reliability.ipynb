# 02 — 4-state Markov-chain reliability

Builds the generator matrix $Q$ from paper eq. (7), computes the stationary distribution and steady-state availability, and simulates one sample path.

```python
import numpy as np
import matplotlib.pyplot as plt

from dcrisk.reliability.markov import (
    MarkovParams, build_Q, stationary_distribution, availability,
    mtbf_from_Q, simulate_chain, STATE_NAMES
)

p = MarkovParams(lam=1/8760, lam_f=1/4380, mu=1/24, mu_r=1/8)
Q = build_Q(p)
print('Q =\n', Q)
```

```python
pi = stationary_distribution(Q)
for name, prob in zip(STATE_NAMES, pi):
    print(f'  pi[{name:<4}] = {prob:.6f}')
print(f'\nSteady-state availability = {availability(Q)*100:.4f}%')
print(f'MTBF (hours from OK to F) = {mtbf_from_Q(Q):,.1f}')
```

```python
rng = np.random.default_rng(7)
times, states = simulate_chain(Q, T_max=8760.0 * 5, rng=rng)

fig, ax = plt.subplots(figsize=(9, 3.5))
ax.step(times, states, where='post', lw=1.4)
ax.set_yticks(range(4)); ax.set_yticklabels(STATE_NAMES)
ax.set_xlabel('time (hours)')
ax.set_title('Sample CTMC path over 5 years')
ax.grid(alpha=0.3)
fig.tight_layout()
```
